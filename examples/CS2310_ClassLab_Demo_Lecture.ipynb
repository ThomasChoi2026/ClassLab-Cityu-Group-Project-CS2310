```c++
#include <iostream>
int main() {
    int a, b;
    std::cin >> a >> b;
    std::cout << a - b << '\n';
}
```

## 1. Predict the output (2 minutes)

The task asks for the **sum** of two integers. The previous cell uses `a - b`, so input `2 3` produces `-1` instead of `5`.

**Discussion:** What do `int a,b;`, `std::cin >> a >> b;`, and `std::cout` each do?

**Import into ClassLab:** Select **Cell 2** and choose **Use this cell as starter code**. Keep the solution cells out of the selection.


## 2. Fix the logic error (3 minutes)

Change `a - b` to `a + b` and submit.

| Standard input | Expected standard output | Purpose |
|---|---|---|
| `2 3` | `5` | Positive numbers |
| `0 0` | `0` | Boundary case |
| `-4 9` | `5` | Mixed signs (hidden test) |

Use the three demo cases listed above. The replay requires an exact match for the supported code and input.


```c++
#include <iostream>
int main() {
    int a, b;
    std::cin >> a >> b;
    std::cout << a + b << '\n';
}
```

## 3. Syntax errors and logic errors (2 minutes)

The earlier program computes the wrong value but has valid syntax. The next cell deliberately omits the semicolon after `std::cin >> a >> b`, causing a compilation error.

The demo shows the **prerecorded compiler diagnostic** for this fixture.


```c++
#include <iostream>
int main() {
    int a, b;
    std::cin >> a >> b
    std::cout << a + b << '\n';
}
```

## 4. Extend the example with procedural abstraction (2 minutes)

Extracting addition into a function makes it easier to reuse, test independently, and organize. Discuss this extension in class. The demo has no replay fixture for it.


```c++
#include <iostream>

int add(int a, int b) {
    return a + b;
}

int main() {
    int a, b;
    if (!(std::cin >> a >> b)) return 1;
    std::cout << add(a, b) << '\n';
}
```

## 5. Live classroom demonstration (3 minutes)

1. Import the notebook, select Cell 2, load the three demo cases, and publish **Fix the addition**.

3. One student keeps `a - b` and two change it to `a + b`. All three submit.
