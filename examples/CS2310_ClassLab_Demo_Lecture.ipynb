# CS2310 ClassLab Demo Lecture｜C++ 输入、逻辑错误与课堂调试

**教学时间：** 12–15 分钟；另附 60 秒项目 Demo。
**教学参考：** 用户提供的教授 Notebook：Lecture 2 *Elements of Computations*、Lecture 3 *Imperative Programming*、Lectures 4–5 *Procedural Abstraction*。本练习为重新编写的合成示例，不是教授原题。

**学习目标：** 理解 `std::cin`、`std::cout`、变量、表达式，以及语法错误与逻辑错误的区别。

**说明：** ClassLab Coursework Demo 的 Run 和提交结果属于**明确标注的预置回放**，并未运行任意学生 C++。课堂加入和提交记录是真实的。


```c++
#include <iostream>
int main() {
    int a, b;
    std::cin >> a >> b;
    std::cout << a - b << '\n';
}
```

## 一、预测输出（2 分钟）

任务要求两个整数的**和**，但上一格用了 `a - b`。对于输入 `2 3`，现有代码会输出 `-1` 而不是 `5`。

**提问：** `int a,b;`、`std::cin >> a >> b;` 和 `std::cout` 各自做什么？

**导入 ClassLab：** 选择 **Cell 2**，点击「填入起始代码」，不要把后面的答案一并组合进去。


## 二、修复逻辑错误（3 分钟）

学生只需要把 `a - b` 改为 `a + b`，然后提交。

| 输入 stdin | 预期 stdout | 用途 |
|---|---|---|
| `2 3` | `5` | 正数测试 |
| `0 0` | `0` | 边界测试 |
| `-4 9` | `5` | 混合符号测试（隐藏用例） |

点击教师页面中的「填入三个演示用例」可以直接生成上述测试；回放仅认可精确匹配的示例代码和输入。


```c++
#include <iostream>
int main() {
    int a, b;
    std::cin >> a >> b;
    std::cout << a + b << '\n';
}
```

## 三、语法错误与逻辑错误（2 分钟）

前面虽然算错，但程序语法合法。下一格故意缺少 `std::cin >> a >> b` 后的分号：这是编译错误，而不是错误答案。

Demo 中相应诊断信息来自固定夹具，**不是真正实时调用编译器**。


```c++
#include <iostream>
int main() {
    int a, b;
    std::cin >> a >> b
    std::cout << a + b << '\n';
}
```

## 四、Procedural Abstraction 延伸（2 分钟）

把加法抽取为函数有助于复用、单独测试和组织程序。本节用于教学讨论，**不属于 Demo 回放支持范围**。


```c++
#include <iostream>

int add(int a, int b) {
    return a + b;
}

int main() {
    int a, b;
    if (!(std::cin >> a >> b)) return 1;
    std::cout << add(a, b) << '\n';
}
```

## 五、教师现场演示（3 分钟）

1. 教师导入 Notebook，选择 Cell 2，发布 **Fix the addition**，加载三个演示用例。
2. 学生使用二维码加入，先点「演示 Run（回放）」，输入 `2 3` 查看标注清晰的预置结果。
3. 一名学生保持 `a - b`，两名改为 `a + b`，分别提交。
4. 教师端展示 3 名学生、2 个通过、1 个不通过，并打开共同修改的 diff。
