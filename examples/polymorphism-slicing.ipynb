### Prevent object slicing

Fix show so a derived animal keeps its dynamic type. Preserve const-correctness and the public interface.


```
#include <iostream>

class Animal {
public:
    virtual const char* voice() const { return "animal"; }
    virtual ~Animal() = default;
};

class Cat final : public Animal {
public:
    const char* voice() const override { return "meow"; }
};
```

```
void show(const Animal animal) {
    std::cout << animal.voice() << '\n';
}
```

```
int main() {
    int kind = 0;
    if (!(std::cin >> kind)) return 0;
    if (kind == 1) {
        Cat cat;
        show(cat);
    } else {
        Animal animal;
        show(animal);
    }
}
```

Teacher note: passing a Cat by value into Animal slices away the derived part. Use a const reference to preserve virtual dispatch.
